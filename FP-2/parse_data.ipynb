# parse_data.ipynb

This notebook parses the data files used for the FP-2 assignment.

<br>
<br>

First let's read the attached data file:

```python
import pandas as pd

df0 = pd.read_csv('Earthquakes.csv')

df0.describe()
```

<br>
<br>

The dependent and independent variables variables (DVs and IVs) that we are interested in are:

**DVs**:
- Death Description

**IVs**:
Mag (magnitude), Focal Depth (km) (depth of earthquake focus in kilometers), Latitude (geographic latitude), Longitude (geographic longitude), Tsu (tsunami indicator), and Vol (volcanic activity indicator)


<br>
<br>

Let's extract the relevant columns:

```python
df = df0[['Death Description', 'Mag', 'Focal Depth (km)', 'Latitude', 'Longitude', 'Tsu', 'Vol']]
df = df.dropna(subset=["Death Description", "Mag", "Focal Depth (km)", "Latitude", "Longitude"])
df.describe()
```

<br>
<br>

Next let's use the `rename` function to give the columns simpler variable names:

```python
df = df.rename( columns={'Focal Depth (km)':'Focal Depth', 'Latitude':'Lat', 'Longitude':'Long'} )

df.describe()
```
